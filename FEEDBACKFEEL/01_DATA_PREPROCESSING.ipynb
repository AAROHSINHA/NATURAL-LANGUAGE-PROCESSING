`KEEPING ONLY THE TEXT AND SENTIMENT COLUMN`
> For our helper function we will do the same. We need to ensure that the data that goes into it is in form [text, label] with these as column names

```python
train_data = train_data[["text", "sentiment"]]
test_data = test_data[["text", "sentiment"]]
train_data.rename(columns={'sentiment': 'label'}, inplace=True)
test_data.rename(columns={'sentiment':'label'}, inplace=True)
train_data.head()
```

```python
# CONVERTING LABEL TO NUMERIC
def encode_labels(value):
    priority = {
        "positive": 1,
        "neutral": 0,
        "negative": -1
    }
    if pd.isna(value):
        return None
    return priority.get(value, None)


train_data["label"] = train_data["label"].apply(encode_labels)
test_data["label"] = test_data["label"].apply(encode_labels)
```

```python
test_data.head()
```

```python
test_data["label"].value_counts()
```

```python
### ***LET'S VISUALIZE SOME DATA SAMPLES***
priority = {
        1: "positive",
        0: "neutral",
        -1: "negative"
    }
def samples(label=1, n=10):
  samples = train_data[train_data["label"] == label]
  samples_shuffled = samples.sample(frac=1).reset_index(drop = True)
  samples_n = list(samples_shuffled["text"].iloc[:n])

  print("---------------------------------------")
  print(priority[label].upper())
  print("---------------------------------------")

  for index, sample in enumerate(samples_n):
    print(f"SAMPLE {index + 1} : {sample}")
    print(" ")

samples(label=-1)
```

## ***DATA PREPROCESSING***
**Now we will basically clean the data and convert them into some useful form to feed to the model**

`THE MAIN STEPS ARE`
* DATA CLEANING
* TOKENIZATION
* STOPWORD REMOVAL
* STEMMING / LEMMATIZATION

```python
# Drop rows with NaN in the 'text' column
train_data = train_data.dropna(subset=["text"]).reset_index(drop=True)
test_data = test_data.dropna(subset=["text"]).reset_index(drop=True)
```

```python
# 1. converting to lowercase
# 2. removing non letters (doesnt add any information)
# 3. removing non whitespaces (last 2 points is basically keeping letters + whitespaces)
def preprocess1(val):
  str = ""
  for v in val:
    if v.isalnum() or v == ' ':
      str += v
  return str

train_data["text"] = train_data["text"].apply(preprocess1)
test_data["text"] = test_data["text"].apply(preprocess1)
```

```python
samples()
```

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
## 2. TOKENIZATION
from nltk.tokenize import word_tokenize
def tokenize(val):
  return word_tokenize(val)

train_data["text"] = train_data["text"].apply(tokenize)
test_data["text"] = test_data["text"].apply(tokenize)
```

```python
train_data.head()
```

```python
## 3. STOPWORD REMOVAL
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
list(stop_words)[:5]
```

```python
stop_words = list(stop_words)
def remove_stopwords(val):
    return [word for word in val if word not in stop_words]

train_data["text"] = train_data["text"].apply(remove_stopwords)
test_data["text"] = test_data["text"].apply(remove_stopwords)
```

```python
train_data.head(5)
```

```python
## 4. STEMMING - PORTER STEMMING
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
def stem_words(val):
  stemmed = [stemmer.stem(word) for word in val]
  return stemmed

train_data["text"] = train_data["text"].apply(stem_words)
test_data["text"] = test_data["text"].apply(stem_words)
```

```python
train_data.head()
```

```python
data = pd.concat([train_data, test_data], axis=0).reset_index(drop=True)
len(data)
```

```python
## 1. BAG OR WORDS
from sklearn.feature_extraction.text import CountVectorizer

data_text = [''.join(tokens) for tokens in data['text']]

# Fit on full dataset
vectorizer = CountVectorizer(max_features=10000)
X = vectorizer.fit_transform(data_text)
X[:5].toarray()
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Join tokens
data_text = [' '.join(tokens) for tokens in data["text"]]

# Fit vectorizer
vectorizer = CountVectorizer(max_features=10000)
X = vectorizer.fit_transform(data_text)

# Convert sparse matrix to DataFrame
data_bow = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
data_bow["label"] = data["label"].values

data_bow.head()
```

```python
from sklearn.decomposition import TruncatedSVD

# Suppose X is your sparse TF-IDF/BoW matrix
svd = TruncatedSVD(n_components=300, random_state=42)
data_bow_pca = svd.fit_transform(X)

print(X.shape)       # (num_samples, vocab_size)
print(data_bow_pca.shape)  # (num_samples, 300)
```

```python
columns = [f"pc{i+1}" for i in range(data_bow_pca.shape[1])]

df_bow_pca = pd.DataFrame(data_bow_pca, columns=columns)
df_bow_pca.head()
```
